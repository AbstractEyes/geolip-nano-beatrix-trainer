# nano Beatrix v1.1 — the mixed-phase arm

[![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/AbstractEyes/geolip-nano-beatrix-trainer/blob/main/colab/nano_v1_1_mixed.ipynb)

Two results of 2026-10-17 and 10-18 shape this arm. The frames-then-carry arm (a frames body continued on carry rows) gained the far read (0.479 of far questions answered whole) and lost the near one (copy 0.885 to 0.531), and the loss fell in the carry phase's first 0.2 billion bytes, before any carry habit had formed: the data form displaces the skill, not the habit. The register fade (the same frame gauges in the math and code registers on every arm's checkpoints) showed that a register's lookup survives a framed phase only if that register is in the mix: the caption-register phases drove the math lookup of the hub bodies to near zero within the first 600 steps while the plain-text controls held or raised it. THE MIXED PHASE puts both forms and all three registers in one phase: a framed row (three rows in four, the bed's dose; the same coin and text position as every other arm's row) is a carry row two times in three and a frames row once, its register drawn caption one time in two, math and code one in four each; the plain rows stay. A caption carry row here is the carry arm's row byte for byte, a caption frames row the frames arm's. The phase is 1.5 billion bytes so that its carry rows total the carry arm's 0.75 billion (the far habit's dose kept) while the frames rows add 0.375 billion.

| craft (one a body) | starts from | its phase |
|---|---|---|
| nano-<body>-v1.1-mx | the body's diet end (step 61,422: the same start as the frames, control and carry arms) | carry rows (0.50 of rows), frames rows (0.25), plain text (0.25, half of it beginning at a document start: DOC and the document's first bytes, so a window-initial DOC no longer announces a frame with certainty), the records' registers caption / math / code at 0.50 / 0.25 / 0.25; 1.5 billion bytes, steps 61,422 to 67,145 |

The reads: the before row (the diet's end), the bed's frame gauges and the carry gauge at 62,000, 64,000, 66,000 and the end, the full eval every 2,000 steps. C4 tabulates the diet's end, the end of this arm, the frames arm, the carry arm and the control; runs the carry gauge on this arm's end beside the carry arm's end; and reads the frame gauges in the math and the code register on this arm's end, the carry arm's end and the control's end.

The pass rule, written before the run: NEAR KEPT when this arm's copy (caption) stays within 0.10 of the frames arm's end; FAR GAINED when its far question whole with its own carries reaches 0.20 or more, FAR MATCHED when it is within 0.10 of the carry arm's; TEXT when its held-out bits sit within 0.01 of the carry arm's end; REGISTERS KEPT when its math and code copy at the end are not below the control's end in that register. All four met: one phase holds the near read, the far habit and every register. NEAR lost: the two forms compete in one mix at this dose, and the frames share rises next. FAR not gained: the carry habit needs the carry rows' full share, and the phase gets longer, not denser. REGISTERS lost with the rest met: the register share rises.

Length: about 40 minutes a body on the card the frames train ran on (27 minutes for a 1.0 billion byte phase), plus the before reads, the carry gauge and six register reads (a minute or two each). One body by default (Pollux dual: the best carry body at 0.59 and the cheapest step); the last cell shows how to run the others. Needs: the body's diet-end state in the repo (C2 checks), and HF_TOKEN in this notebook's secrets with write access to AbstractPhil/alephllm-nano-castor-pollux.


```python
# C1 — pinned installs + THE DECISION BLOCK. RESTART RUNTIME if the version line below changes.
import os, sys
os.environ['HF_HUB_DISABLE_PROGRESS_BARS'] = '1'   # a progress bar is not allowed to cost a session
%pip install -q "geolip-nano-beatrix-trainer[corpus] @ git+https://github.com/AbstractEyes/geolip-nano-beatrix-trainer@v0.1.52"
import torch
import geolip.nano_beatrix as N
import geolip.alephllm as A
from geolip.nano_beatrix import nano_v1 as V1
print('geolip.nano_beatrix', N.__version__, '| geolip.alephllm', A.__version__, '| torch', torch.__version__,
      '| card', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'NONE')
assert tuple(int(x) for x in N.__version__.split('.')) >= (0, 1, 52), (
    'RESTART REQUIRED: the runtime still holds an older package — Runtime > Restart, then rerun from C1')
assert torch.cuda.is_available(), 'no card: Runtime > Change runtime type > GPU'

def _token():
    try:
        from google.colab import userdata
        return userdata.get('HF_TOKEN')
    except Exception:
        return os.environ.get('HF_TOKEN')
HF_TOKEN = _token()                                   # the value is never printed
assert HF_TOKEN, 'HF_TOKEN missing: the start states are read from the repo and the continuations ship to it'
os.environ['HF_TOKEN'] = HF_TOKEN

# ============ THE DECISION BLOCK — every setting from the record, with its source beside it ============
OUT = '/content/nano_v1_1_mixed'                      # the session folder: the corpus, the runs, the bed's gauge folder
REPO = V1.REPO                                        # AbstractPhil/alephllm-nano-castor-pollux: the v1.1 folders read, the new ones written
NAMES = ['nano-pollux-dual-v1.1-mx']                  # one body first (the best carry body); all four: list(V1.CRAFTS_MIX); the second seed: ['nano-pollux-dual-v1.1-mx2']
MB, GA = 16, 4                                        # the 2s shape, the fastest on the v1.1 benchmark (262,144 bytes a step)
GAUGE_EVERY = 2000                                    # the bed's gauges at 62,000, 64,000 and 66,000 (every full eval) and at the end
CORPUS_BYTES = 1.0e9                                  # the frames corpus (fineweb-edu, the held-out head skipped), built once in C2
MAX_HOURS = None                                      # this session's budget in hours (None: until every craft ends or the cell is stopped)
print('decision block:', len(NAMES), 'crafts:', ', '.join(NAMES), '| the phase', V1.CONT_TOKENS_OF['mx'] / 1e9, 'B bytes |', MB, 'x', GA,
      '| gauges every', GAUGE_EVERY, 'steps | corpus', CORPUS_BYTES / 1e9, 'GB | ships to', REPO)
```

```python
# C2 (P) — PREFLIGHT: every body's diet-end state in the repo, the held-out head, the frames corpus (built once; about a minute or two
# a gigabyte from the stream), the mixed rows' composition pace against the card's, the push probe. Nothing trains here.
import json, time
from huggingface_hub import HfApi
from geolip.nano_beatrix.bed_corpus import write_heldout, build as build_corpus
from geolip.nano_beatrix import frames_data as FD
api = HfApi(token=HF_TOKEN)
files = set(api.list_repo_files(REPO))
missing = []
for n in NAMES:
    spec = V1.cont_init(n, MB * GA * V1.CTX)
    path = f"{spec['prefix']}/{spec['payload']}"
    if path not in files:
        missing.append(path)
assert not missing, f'not every body has its diet-end state: missing {sorted(set(missing))}'
print('start states: every body has its diet-end state at step', spec['step'])
held = write_heldout(f'{OUT}/bed')
folder = f'{OUT}/frames_corpus'
if not os.path.exists(f'{folder}/bed_corpus.bin'):
    build_corpus(folder, int(CORPUS_BYTES))
packed = FD.load_corpus(folder)
n, t0, kinds = 0, time.time(), {}
while time.time() - t0 < 10:
    FD.row_bytes(packed, 1337, n, FD.P_FRAMED, FD.MIXED, carry='mixed')
    k = FD.mixed_draw(__import__('numpy').random.default_rng([1337, n, 2]))
    kinds[k] = kinds.get(k, 0) + 1
    n += 1
rate = n / (time.time() - t0)
need = 879_000 / FD.ROW_BYTES                         # the fastest craft's rows a second (Castor, 879k bytes a second)
print(f'frames corpus {len(packed) / 1e9:.2f} GB; the mixed rows compose at {rate:,.0f} a second against {need:.0f} needed'
      + ('' if rate > 1.5 * need else ' — TOO SLOW: the card will wait on the data'))
print('kinds and registers drawn over', n, 'rows:', dict(sorted(kinds.items())))

PUSH_OK = False
path = 'preflight/push_probe_v1_1_mixed.json'        # its own probe file
body = json.dumps({'utc': time.strftime('%Y-%m-%dT%H:%M:%SZ', time.gmtime()), 'card': torch.cuda.get_device_name(0),
                   'compose_rows_s': rate, 'crafts': NAMES}).encode()
api.upload_file(path_or_fileobj=body, path_in_repo=path, repo_id=REPO, commit_message='nano v1.1 mixed phase: push probe')
PUSH_OK = path in set(api.list_repo_files(REPO))
print('push probe', 'READ BACK OK' if PUSH_OK else 'NOT VISIBLE after upload', path)

# the gauge's smoke on this card before any training: the per-row dump on 8 rows of the start state (about a minute; a body at the
# diet's end writes no carries, so the far question reads 0.00 here)
src = V1.ALL[NAMES[0]]['source']
spec = V1.cont_init(NAMES[0], MB * GA * V1.CTX)            # 0.1.50: the first craft's own start (a list may mix arms that start at different steps)
smoke = V1.gauge_stored(src, spec['step'], OUT, repo=REPO, carry=True, token=HF_TOKEN, extra=dict(carry_rows=8, carry_dump=1))
c = smoke['carry']
rows8 = sum(1 for _ in open(f"{OUT}/bed/{c['dump']['jsonl']}", encoding='utf-8'))
assert rows8 == c['n'] == 8, (rows8, c['n'])
print(f"gauge smoke on {src} at {spec['step']:,}: {rows8} rows dumped; carries a row {c['carries_per_row']:.2f}; "
      f"far question whole {c['full_value']:.2f}")
```

```python
# C3 (T) — THE CONTINUATIONS: each craft starts from its body's diet-end state (its first row: the before, read by the gauges), then
# runs the mixed phase; the trainer's lines and the bed's gauge lines ([v1]) print as they come. Rerun the cell to continue (a new
# session too). The cell's stop button saves the running craft's state before it stops.
assert PUSH_OK, 'the push probe did not read back: nothing would ship'
HIST = V1.run(OUT, HF_TOKEN, NAMES, micro_batch=MB, grad_accum=GA, repo=REPO, gauge_every=GAUGE_EVERY, max_hours=MAX_HOURS,
              frames_corpus=f'{OUT}/frames_corpus')
from IPython.display import Markdown, display
display(Markdown(V1.mixed_table(HIST)))
```

```python
# C4 — THE FIVE-COLUMN READ, READING WITH NOTES, THE REGISTERS, THE MARKS: the boundary rows of this body's arms from the repo (frames,
# control, carry, mixed), the table (the diet's end = the start, after, the frames arm, the carry arm, the control, the differences); the
# carry gauge on the mixed craft's end beside the carry arm's end (192 rows each; gauge_stored pulls the final and runs the gauge here, a
# few minutes a read); the frame gauges in the math and the code register on the mixed end, the carry arm's end and the control's end;
# then the marks. Run any time; it trains nothing.
import json
from IPython.display import Markdown, display
from huggingface_hub import snapshot_download
BODIES = sorted({V1.body_of(n) for n in NAMES})      # nano-<body>-v1.1
MXC = {V1.body_of(n): n for n in NAMES}               # the mixed craft read for each body: -mx, or -mx2 (the second seed, 0.1.52)
ALL_ARMS = [f'{b}-{a}' for b in BODIES for a in ('fr', 'ct', 'cf', 'mx')] + [n for n in NAMES if not n.endswith('-mx')]
hub = snapshot_download(REPO, token=HF_TOKEN, local_dir=f'{OUT}/hub',
                        allow_patterns=[f'{n}/runs/*' for n in ALL_ARMS] + [f'{n}/v1_boundaries.json' for n in ALL_ARMS])
hist = {}
for n in ALL_ARMS:
    p = f'{hub}/{n}/v1_boundaries.json'
    hist[n] = json.load(open(p, encoding='utf-8')) if os.path.exists(p) else []
display(Markdown(V1.mixed_table(hist)))
TPS = MB * GA * V1.CTX

def end_row(n):
    e = hist.get(n) or []
    return e[-1] if e and e[-1].get('phase') == V1.CONT_PHASE[V1.ALL[n]['cont']][0] else None

def bed(n, k):
    e = end_row(n)
    return None if e is None else (e.get('bed') or {}).get(k)

def f(x, nd=3):
    return '-' if x is None else f'{x:.{nd}f}'
rows = ['| body | craft | end step | whole value at the far question, own carries | first byte | value bits | carries a row | rows with a carry | exact copies | open probability at the grid | distance to the latest definition |',
        '|---|---|---|---|---|---|---|---|---|---|---|']
carry_read = {}
for b in BODIES:
    for n in (MXC[b], f'{b}-cf'):
        arm = n[len(b) + 1:]
        end = V1.plan_steps(n, TPS)                      # 67,145 for the mixed craft, 65,237 for the carry arm
        if end_row(n) is None:
            rows.append(f'| {b} | {arm} | {end:,} | not finished | | | | | | | |')
            continue
        c = V1.gauge_stored(n, end, OUT, repo=REPO, carry=True, token=HF_TOKEN).get('carry') or {}
        carry_read[n] = c
        rows.append(f"| {b} | {arm} | {end:,} | {f(c.get('full_value'))} | {f(c.get('first_byte'))} | {f(c.get('value_bits'), 1)} | "
                    f"{f(c.get('carries_per_row'), 2)} | {f(c.get('rows_with_carries'), 2)} | {f(c.get('carry_exact'), 2)} | "
                    f"{f(c.get('open_prob'))} | {f(c.get('dist_latest'), 0)} |")
display(Markdown('\n'.join(rows)))

# the registers: the same frame gauges in the math and the code register (the training phases framed the caption register only, except
# this arm's); the control's end is the bar
reg_rows = ['| body | craft | register | copy | nested | redefined key | 1,024-byte frame | far question, own carries | carries a row | held-out bits |',
            '|---|---|---|---|---|---|---|---|---|---|']
reg_read = {}
for b in BODIES:
    for n in (MXC[b], f'{b}-cf', f'{b}-ct'):
        arm = n[len(b) + 1:]
        end = V1.plan_steps(n, TPS)
        if end_row(n) is None:
            continue
        for reg in ('math', 'code'):
            sc = V1.gauge_scalars(V1.gauge_stored(n, end, OUT, repo=REPO, carry=True, token=HF_TOKEN, extra=dict(register=reg)))
            reg_read[(n, reg)] = sc
            reg_rows.append(f"| {b} | {arm} | {reg} | {f(sc.get('bed/frame/copy_whole'))} | {f(sc.get('bed/frame/nested_whole'))} | "
                            f"{f(sc.get('bed/frame/redefined_key_whole'))} | {f(sc.get('bed/scaling/frame_1024_whole'))} | "
                            f"{f(sc.get('bed/carry/full_value'))} | {f(sc.get('bed/carry/carries_per_row'), 2)} | {f(sc.get('bed/text/heldout_bits'), 4)} |")
display(Markdown('\n'.join(reg_rows)))

# the marks, by the rule written before the run
marks = ['| body | mark | rule | value | verdict |', '|---|---|---|---|---|']
for b in BODIES:
    fr, cf, ct = (f'{b}-{a}' for a in ('fr', 'cf', 'ct'))
    mx = MXC[b]
    if end_row(mx) is None:
        marks.append(f'| {b} | all | | the mixed arm has not finished | |')
        continue
    near, near_fr = bed(mx, 'bed/frame/copy_whole'), bed(fr, 'bed/frame/copy_whole')
    far, far_cf = (carry_read.get(mx) or {}).get('full_value'), (carry_read.get(cf) or {}).get('full_value')
    txt, txt_cf = bed(mx, 'bed/text/heldout_bits'), bed(cf, 'bed/text/heldout_bits')
    ok = lambda c: 'MET' if c else 'NOT MET'
    if near is not None and near_fr is not None:
        marks.append(f'| {b} | NEAR KEPT | copy within 0.10 of the frames arm end | {near:.3f} against {near_fr:.3f} ({near - near_fr:+.3f}) | {ok(near >= near_fr - 0.10)} |')
    if far is not None:
        marks.append(f'| {b} | FAR GAINED | far question whole 0.20 or more | {far:.3f} | {ok(far >= 0.20)} |')
        if far_cf is not None:
            marks.append(f'| {b} | FAR MATCHED | within 0.10 of the carry arm | {far:.3f} against {far_cf:.3f} ({far - far_cf:+.3f}) | {ok(far >= far_cf - 0.10)} |')
    if txt is not None and txt_cf is not None:
        marks.append(f'| {b} | TEXT | held-out bits within 0.01 of the carry arm end | {txt:.4f} against {txt_cf:.4f} ({txt - txt_cf:+.4f}) | {ok(abs(txt - txt_cf) <= 0.01)} |')
    for reg in ('math', 'code'):
        a, c = reg_read.get((mx, reg), {}).get('bed/frame/copy_whole'), reg_read.get((ct, reg), {}).get('bed/frame/copy_whole')
        if a is not None and c is not None:
            marks.append(f'| {b} | REGISTERS KEPT ({reg}) | copy not below the control end | {a:.3f} against {c:.3f} ({a - c:+.3f}) | {ok(a >= c)} |')
display(Markdown('\n'.join(marks)))
display(Markdown(V1.table(hist)))
%load_ext tensorboard
%tensorboard --logdir {hub}
```

```python
# C4b — SHIP THE READS: C4's tables live only in this notebook's output; this cell writes them as one JSON file and uploads it under
# the first craft's prefix (<craft>/reads/mixed_reads_<stamp>.json) so the record is read from the Hub like every other result. Run
# after C4; it trains nothing.
import json, time
from huggingface_hub import HfApi
stamp = time.strftime('%Y%m%d_%H%M')
rec = dict(stamp=stamp, names=NAMES, carry_read=carry_read, reg_read={n + '|' + r: v for (n, r), v in reg_read.items()})
p = f'{OUT}/mixed_reads_{stamp}.json'
json.dump(rec, open(p, 'w', encoding='utf-8'), indent=1)
dest = f'{NAMES[0]}/reads/mixed_reads_{stamp}.json'
HfApi(token=HF_TOKEN).upload_file(path_or_fileobj=p, path_in_repo=dest, repo_id=REPO,
                                  commit_message='the mixed arm post-hoc reads: the carry gauge and the register gauges (notebook C4)')
print('shipped', dest)
```

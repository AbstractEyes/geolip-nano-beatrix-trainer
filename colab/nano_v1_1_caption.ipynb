# nano Beatrix v1.1 — the caption phase, the frameless question

[![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/AbstractEyes/geolip-nano-beatrix-trainer/blob/main/colab/nano_v1_1_caption.ipynb)

Three results of 2026-10-18 and 10-19 shape this phase. The mixed arm holds every read in one phase (the near read, the far habit, the three registers, the text). The gate screen on the four mixed and note bodies found no body that proposes notes which pay for the text after them (NOT READY on both proposers). On the diffusion side the framed nano arms, read as sliders, move the picture on the wordings they were trained on and not on new wordings of the same meaning: the frames taught the format, not the meaning of the words. THE CAPTION PHASE puts real captions behind real records: the Beatrix caption arm's caption pack (photograph rows with their attributes in real wordings, illustration rows with their tags, web-image and COCO rows with their captions). A row's attributes become records whose keys are PATHS of a closed vocabulary, the system then the attribute (`look hair`, `look eyes`, `person gender`, `wear clothing`, `scene mood`, `shot light`, `thing <name>`), and a document's records are written as one sized frame in which every system's records sit in a child frame of their own, the children nested inside the document's frame at shuffled places: a symbolic trie, a system of systems. A caption document is the caption head, the frame (a shuffled subset of the row's records, one to eight), a newline, ONE of the row's several captions (the same attributes said four ways across the epochs) and, in half the documents, the question of a framed record; a caption window is documents back to back. The phase keeps the mixed arm's coin and plain rows; of its framed rows a third are carry rows (the far habit, the three registers) and two thirds are caption windows.

THE FIRST RUN (2026-10-19 02:4x to 04:2x, Pollux dual, trainer 0.1.55) taught the format and not the meaning. After the phase the body copies a record out of the nested system frames whole in 0.961 of 256 held items (the start 0.004: the flat-form skill did not transfer to the trie form), but asked from the caption alone it answers 0.168, and the control, whose frames never matched their captions, answers 0.172 with the same value bits (7.20 against 7.23): each key's commonest value, not the caption (the age is always 25 to 35 and both answer it; the colour words come out plausible and wrong). The cause is in the data form: the frame holding the answer stood before every question, so copying the frame answered it and the caption was never needed. THE REPAIR in this version: the frame is withheld in half the caption documents, and a frameless document always asks its question, a record of its own, so the caption has to answer it; the control keeps the swap (its frame and its question from another document), so in its frameless documents the question's truth is unrelated to the caption and only the prior can be learned; and every extraction number is read beside the prior-only baseline, each key's commonest training value scored on the same held items.

| craft | starts from | its phase |
|---|---|---|
| nano-<body>-v1.1-mx-cp2 | the body's mixed arm's end (step 67,145) | caption windows (0.50 of rows; the frame withheld in half the documents, which then always ask), carry rows (0.25), plain text (0.25, half of it at a document start); 1.5 billion bytes, steps 67,145 to 72,867 |
| nano-pollux-dual-v1.1-mx-cpx2 (the control) | the same | the same phase with every document's frame and question taken from ANOTHER document: in a frameless document the question's truth has nothing to do with the caption |

The reads: the before row (the mixed arm's end), the bed's frame gauges and the carry gauge every 2,000 steps and at the end, and on the pack's HELD split (rows never trained) the reads run by C4 on the start, the end and the control's end: THE EXTRACTION READ (the caption alone, then the question of an attribute the caption states: the value), THE COPY READ (the frame, the caption, the question) and THE PRIOR (each key's commonest training value on the same items).

The pass rule, written before the run: NEW WORDINGS when the caption end's extraction read answers 0.20 or more of the held questions whole above BOTH the control's and the prior; COPY when its copy read is within 0.10 of the first run's 0.961; TEXT KEPT when its held-out bits sit within 0.01 of the start; NEAR AND FAR KEPT when its bed copy and its far question whole sit within 0.10 of the start; THE CONTROL FAILS when the control's extraction read sits within 0.05 of the prior. All met: the body reads the meaning of the words from the caption into the records once the frame stops answering for it. NEW WORDINGS missed with THE CONTROL FAILS met: the caption reading needs more dose or more frameless documents, not another form. THE CONTROL above the prior by 0.05 or more: the read leaks (the captions name the value in a way the question gives away) and the read is rebuilt before any claim. NEAR KEPT is expected to fail as in the first run (the old frame form is not in these rows) and is reported, not judged. The gate read on caption documents follows in a later version.

Length: about 40 minutes a craft on the card the first run used (an RTX PRO 6000), plus the pack (a few minutes: the first shards of each source, about a gigabyte, built once) and the reads (a minute or two each). Pollux dual's repaired caption craft and its control by default; the last cell shows how to run the other bodies. Needs: the body's mixed arm's end in the repo (C2 checks), and HF_TOKEN in this notebook's secrets with write access to AbstractPhil/alephllm-nano-castor-pollux.


```python
# C1 — pinned installs + THE DECISION BLOCK. RESTART RUNTIME if the version line below changes.
import os, sys
os.environ['HF_HUB_DISABLE_PROGRESS_BARS'] = '1'   # a progress bar is not allowed to cost a session
%pip install -q "geolip-nano-beatrix-trainer[corpus] @ git+https://github.com/AbstractEyes/geolip-nano-beatrix-trainer@v0.1.56"
import torch
import geolip.nano_beatrix as N
import geolip.alephllm as A
from geolip.nano_beatrix import nano_v1 as V1
print('geolip.nano_beatrix', N.__version__, '| geolip.alephllm', A.__version__, '| torch', torch.__version__,
      '| card', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'NONE')
assert tuple(int(x) for x in N.__version__.split('.')) >= (0, 1, 56), (
    'RESTART REQUIRED: the runtime still holds an older package — Runtime > Restart, then rerun from C1')
assert torch.cuda.is_available(), 'no card: Runtime > Change runtime type > GPU'

def _token():
    try:
        from google.colab import userdata
        return userdata.get('HF_TOKEN')
    except Exception:
        return os.environ.get('HF_TOKEN')
HF_TOKEN = _token()                                   # the value is never printed
assert HF_TOKEN, 'HF_TOKEN missing: the start states are read from the repo and the continuations ship to it'
os.environ['HF_TOKEN'] = HF_TOKEN

# ============ THE DECISION BLOCK — every setting from the record, with its source beside it ============
OUT = '/content/nano_v1_1_caption'                    # the session folder: the corpus, the pack, the runs, the bed's gauge folder
REPO = V1.REPO                                        # AbstractPhil/alephllm-nano-castor-pollux: the mixed arms read, the new crafts written
NAMES = ['nano-pollux-dual-v1.1-mx-cp2', 'nano-pollux-dual-v1.1-mx-cpx2']   # the repaired caption craft (frameless questions) and its control on the best carry body; all four bodies: list(V1.CRAFTS_CAPTION2)
ARM = V1.ALL[NAMES[0]]['cont']                        # cp2 (the first form again: NAMES of -mx-cp and -mx-cpx give cp)
CTRL = ARM.replace('cp', 'cpx', 1)                    # the control's arm code (cpx2)
MB, GA = 16, 4                                        # the 2s shape, the fastest on the v1.1 benchmark (262,144 bytes a step)
GAUGE_EVERY = 2000                                    # the bed's gauges every full eval and at the end
CORPUS_BYTES = 1.0e9                                  # the frames corpus (fineweb-edu, the held-out head skipped) for the carry rows and the plain rows, built once in C2
PACK_ROWS = None                                      # the caption pack: None = the first shards of each source whole (about a gigabyte); a number = that many rows a source (a smoke)
MAX_HOURS = None                                      # this session's budget in hours (None: until every craft ends or the cell is stopped)
PACK = f'{OUT}/caption_pack'
print('decision block:', len(NAMES), 'crafts:', ', '.join(NAMES), '| the phase', V1.CONT_TOKENS_OF[ARM] / 1e9, 'B bytes |', MB, 'x', GA,
      '| gauges every', GAUGE_EVERY, 'steps | corpus', CORPUS_BYTES / 1e9, 'GB | pack rows', PACK_ROWS or 'whole shards', '| ships to', REPO)
```

```python
# C2 (P) — PREFLIGHT: every craft's start state in the repo, the held-out head, the frames corpus (built once), THE CAPTION PACK (fetched
# and built once; its counts), one composed caption window decoded, the caption rows' composition pace against the card's, the push
# probe, the caption read's smoke on the start state. Nothing trains here.
import json, time
import numpy as np
from huggingface_hub import HfApi
from geolip.nano_beatrix.bed_corpus import write_heldout, build as build_corpus
from geolip.nano_beatrix import frames_data as FD
from geolip.nano_beatrix import caption_pack as CP
from geolip.alephllm.data import special_tokens as ST
api = HfApi(token=HF_TOKEN)
files = set(api.list_repo_files(REPO))
missing = []
for n in NAMES:
    spec = V1.cont_init(n, MB * GA * V1.CTX)
    path = f"{spec['prefix']}/{spec['payload']}"
    if path not in files:
        missing.append(path)
assert not missing, f'not every craft has its start state: missing {sorted(set(missing))}'
print('start states: every craft starts from its mixed arm at step', spec['step'])
held = write_heldout(f'{OUT}/bed')
folder = f'{OUT}/frames_corpus'
if not os.path.exists(f'{folder}/bed_corpus.bin'):
    build_corpus(folder, int(CORPUS_BYTES))
packed = FD.load_corpus(folder)

stats = CP.ensure(PACK, HF_TOKEN, PACK_ROWS, log=print)
PK = CP.load(PACK)
HELD = CP.load(PACK, 'held')
print(f"the pack: {PK.n:,} training documents by kind {dict((k, len(v)) for k, v in PK.by_kind.items())}; {HELD.n:,} held; "
      f"{sum(1 for i in range(HELD.n) if HELD.records(i)):,} held documents with records")
NAMES_ST = {v: k for k, v in vars(ST).items() if isinstance(v, int) and 0xF0 <= v <= 0xFF and k.isupper()}
def show(b):
    return ''.join(f'<{NAMES_ST[c]}>' if c in NAMES_ST else chr(c) if 32 <= c < 127 or c == 10 else f'\\x{c:02x}' for c in b)
FORM = {'cp': FD.CAPTION, 'cpx': FD.CAPTION_CTRL, 'cp2': FD.CAPTION2, 'cpx2': FD.CAPTION2_CTRL}[ARM]
row, metas = FD.caption_window(np.random.default_rng(1), PK, FORM)
print(f'one caption window of the {ARM} form: {len(metas)} documents, {sum(1 for m in metas if not m.get("framed", True))} frameless; the first:', metas[0])
print(show(bytes(row[:900])))

n, t0, kinds = 0, time.time(), {}
while time.time() - t0 < 10:
    FD.row_bytes(packed, 1337, n, FD.P_FRAMED, FORM, carry='caption' if ARM == 'cp' else 'caption_ctrl' if ARM == 'cpx' else 'caption2' if ARM == 'cp2' else 'caption2_ctrl', pack=PK)
    k = FD.mixed_draw(np.random.default_rng([1337, n, 2]), FORM)
    kinds[k] = kinds.get(k, 0) + 1
    n += 1
rate = n / (time.time() - t0)
need = 879_000 / FD.ROW_BYTES                         # the fastest craft's rows a second (Castor, 879k bytes a second)
print(f'frames corpus {len(packed) / 1e9:.2f} GB; the caption rows compose at {rate:,.0f} a second against {need:.0f} needed'
      + ('' if rate > 1.5 * need else ' — TOO SLOW: the card will wait on the data'))
print('kinds and registers drawn over', n, 'rows:', dict(sorted(kinds.items())))

PUSH_OK = False
path = 'preflight/push_probe_v1_1_caption.json'      # its own probe file
body = json.dumps({'utc': time.strftime('%Y-%m-%dT%H:%M:%SZ', time.gmtime()), 'card': torch.cuda.get_device_name(0),
                   'compose_rows_s': rate, 'crafts': NAMES, 'pack_docs': PK.n, 'pack_held': HELD.n}).encode()
api.upload_file(path_or_fileobj=body, path_in_repo=path, repo_id=REPO, commit_message='nano v1.1 caption phase: push probe')
PUSH_OK = path in set(api.list_repo_files(REPO))
print('push probe', 'READ BACK OK' if PUSH_OK else 'NOT VISIBLE after upload', path)

# the caption read's smoke on this card before any training: 8 held items on the start state (the mixed arm's end; it never saw a
# path key, so the extraction read is expected near 0 here and the copy read whatever the near skill gives on real records)
src = V1.ALL[NAMES[0]]['source']
spec = V1.cont_init(NAMES[0], MB * GA * V1.CTX)
smoke = V1.caption_stored(src, spec['step'], f'{OUT}/smoke', repo=REPO, token=HF_TOKEN, rows=8, pack=PACK)
print(f"caption read smoke on {src} at {spec['step']:,}: {smoke['n']} items; extraction whole {smoke['extraction']['whole']:.2f}, "
      f"copy whole {smoke['copy']['whole']:.2f}, the prior {smoke['prior']['whole']:.2f} ({smoke['prior']['n_docs']:,} training documents tallied); "
      f"{smoke['extraction']['seconds'] + smoke['copy']['seconds']:.0f} s")
```

```python
# C3 (T) — THE CONTINUATIONS: each craft starts from its mixed arm's end (its first row: the before, read by the gauges), then runs the
# caption phase; the trainer's lines and the bed's gauge lines ([v1]) print as they come. Rerun the cell to continue (a new session
# too). The cell's stop button saves the running craft's state before it stops.
assert PUSH_OK, 'the push probe did not read back: nothing would ship'
HIST = V1.run(OUT, HF_TOKEN, NAMES, micro_batch=MB, grad_accum=GA, repo=REPO, gauge_every=GAUGE_EVERY, max_hours=MAX_HOURS,
              frames_corpus=f'{OUT}/frames_corpus', caption_pack=PACK, caption_pack_rows=PACK_ROWS)
from IPython.display import Markdown, display
display(Markdown(V1.caption_table(HIST, ARM)))
```

```python
# C4 — THE READS AND THE MARKS: the boundary rows of this body's crafts from the repo (the mixed arm, the caption craft, the control), the
# table (the start, after, the control, the differences); the two held-split reads (256 items each; caption_stored pulls the weights
# and runs the read here, a minute or two a read) on the start, the caption end and the control's end; the reads by key on the
# caption end; then the marks. Run any time; it trains nothing.
import json
from IPython.display import Markdown, display
from huggingface_hub import snapshot_download
BODIES = sorted({V1.body_of(n) for n in NAMES})      # nano-<body>-v1.1
ARMS = [f'{b}-{a}' for b in BODIES for a in ('mx', f'mx-{ARM}', f'mx-{CTRL}')]
hub = snapshot_download(REPO, token=HF_TOKEN, local_dir=f'{OUT}/hub',
                        allow_patterns=[f'{n}/runs/*' for n in ARMS] + [f'{n}/v1_boundaries.json' for n in ARMS])
hist = {}
for n in ARMS:
    p = f'{hub}/{n}/v1_boundaries.json'
    hist[n] = json.load(open(p, encoding='utf-8')) if os.path.exists(p) else []
display(Markdown(V1.caption_table(hist, ARM)))
TPS = MB * GA * V1.CTX

def end_row(n):
    e = hist.get(n) or []
    return e[-1] if e and e[-1].get('phase') == V1.CONT_PHASE[V1.ALL[n]['cont']][0] else None

def bed(n, k):
    e = end_row(n)
    return None if e is None else (e.get('bed') or {}).get(k)

def f(x, nd=3):
    return '-' if x is None else f'{x:.{nd}f}'
reads = {}
rows = ['| body | craft | step | extraction whole | first byte | value bits | the prior | copy whole | first byte | value bits | items | by kind (extraction whole) |',
        '|---|---|---|---|---|---|---|---|---|---|---|---|']
for b in BODIES:
    for n, label in ((f'{b}-mx', 'the start (mixed end)'), (f'{b}-mx-{ARM}', 'caption'), (f'{b}-mx-{CTRL}', 'control')):
        if n not in V1.ALL or (n != f'{b}-mx' and end_row(n) is None):
            if n in V1.ALL and n in NAMES:
                rows.append(f'| {b} | {label} | | not finished | | | | | | | | |')
            continue
        step = V1.plan_steps(n, TPS)
        r = V1.caption_stored(n, step, OUT, repo=REPO, token=HF_TOKEN, rows=256, pack=PACK)
        reads[n] = r
        ex, cp_ = r['extraction'], r['copy']
        kinds = ', '.join(f"{k} {v['whole']:.2f}" for k, v in sorted(ex['by_kind'].items()))
        rows.append(f"| {b} | {label} | {step:,} | {f(ex['whole'])} | {f(ex['first_byte'])} | {f(ex['value_bits'], 1)} | {f((r.get('prior') or {}).get('whole'))} | "
                    f"{f(cp_['whole'])} | {f(cp_['first_byte'])} | {f(cp_['value_bits'], 1)} | {r['n']} | {kinds} |")
display(Markdown('\n'.join(rows)))

for b in BODIES:
    n = f'{b}-mx-{ARM}'
    if n not in reads:
        continue
    ex, cp_ = reads[n]['extraction']['by_key'], reads[n]['copy']['by_key']
    exc = (reads.get(f'{b}-mx-{CTRL}') or {}).get('extraction', {}).get('by_key', {})
    pr = (reads[n].get('prior') or {}).get('by_key', {})
    krows = [f'| key ({b}, the caption end) | items | extraction whole | the control | the prior | copy whole |', '|---|---|---|---|---|---|']
    for k in sorted(ex, key=lambda k: -ex[k]['n']):
        krows.append(f"| {k} | {ex[k]['n']} | {f(ex[k]['whole'])} | {f(exc.get(k, {}).get('whole'))} | {f(pr.get(k, {}).get('whole'))} | {f(cp_.get(k, {}).get('whole'))} |")
    display(Markdown('\n'.join(krows)))
    exs = reads[n]['extraction']['examples'][:12]               # the record's examples: kind, key, value (the truth), read (what the body wrote), right
    display(Markdown('\n'.join(['| kind (the caption end) | question | truth | written | right |', '|---|---|---|---|---|']
                               + [f"| {e['kind']} | {e['key']} | {e['value'].replace('|', '/')} | {e['read'].replace('|', '/')} | {'yes' if e['right'] else 'no'} |" for e in exs])))

# the marks, by the rule written before the run
marks = ['| body | mark | rule | value | verdict |', '|---|---|---|---|---|']
ok = lambda c: 'MET' if c else 'NOT MET'
for b in BODIES:
    mx, cp_, cpx = f'{b}-mx', f'{b}-mx-{ARM}', f'{b}-mx-{CTRL}'
    if end_row(cp_) is None:
        marks.append(f'| {b} | all | | the caption craft has not finished | |')
        continue
    e0, e1, e2 = (reads.get(n) for n in (mx, cp_, cpx))
    if e0 and e1:
        p1 = (e1.get('prior') or {}).get('whole') or 0.0
        x2 = (e2 or {}).get('extraction', {}).get('whole') if e2 else None
        bar = max(x2 or 0.0, p1)
        marks.append(f"| {b} | NEW WORDINGS | extraction whole 0.20 or more above both the control's and the prior | {e1['extraction']['whole']:.3f} against the control's {f(x2)} and the prior {p1:.3f} (the start {e0['extraction']['whole']:.3f}) | {ok(e1['extraction']['whole'] >= bar + 0.20)} |")
        marks.append(f"| {b} | COPY | copy read within 0.10 of the first run's 0.961 | {e1['copy']['whole']:.3f} (the start {e0['copy']['whole']:.3f}) | {ok(e1['copy']['whole'] >= 0.861)} |")
    start = next((e for e in hist.get(cp_) or [] if e.get('phase') == V1.CONT_PHASE['mx'][0]), None) or end_row(mx)
    s = (start or {}).get('bed') or {}
    for mark, k, bar, nd in (('TEXT KEPT', 'bed/text/heldout_bits', 0.01, 4), ('NEAR KEPT', 'bed/frame/copy_whole', 0.10, 3), ('FAR KEPT', 'bed/carry/full_value', 0.10, 3)):
        a, z = bed(cp_, k), s.get(k)
        if a is not None and z is not None:
            cond = abs(a - z) <= bar if mark == 'TEXT KEPT' else a >= z - bar
            marks.append(f'| {b} | {mark} | within {bar} of the start | {a:.{nd}f} against {z:.{nd}f} ({a - z:+.{nd}f}) | {ok(cond)} |')
    if e2:
        p2 = (e2.get('prior') or {}).get('whole') or 0.0
        marks.append(f"| {b} | THE CONTROL FAILS | the control's extraction whole within 0.05 of the prior | {e2['extraction']['whole']:.3f} against the prior {p2:.3f} ({e2['extraction']['whole'] - p2:+.3f}) | {ok(abs(e2['extraction']['whole'] - p2) <= 0.05)} |")
display(Markdown('\n'.join(marks)))
display(Markdown(V1.table(hist)))
%load_ext tensorboard
%tensorboard --logdir {hub}
```

```python
# C4b — SHIP THE READS: C4's tables live only in this notebook's output; this cell writes them as one JSON file and uploads it under
# the first craft's prefix (<craft>/reads/caption_reads_<stamp>.json) so the record is read from the Hub like every other result. Run
# after C4; it trains nothing.
import json, time
from huggingface_hub import HfApi
stamp = time.strftime('%Y%m%d_%H%M')
rec = dict(stamp=stamp, names=NAMES, pack_rows=PACK_ROWS, reads=reads, marks=marks)
p = f'{OUT}/caption_reads_{stamp}.json'
json.dump(rec, open(p, 'w', encoding='utf-8'), indent=1)
dest = f'{NAMES[0]}/reads/caption_reads_{stamp}.json'
HfApi(token=HF_TOKEN).upload_file(path_or_fileobj=p, path_in_repo=dest, repo_id=REPO,
                                  commit_message='the caption phase post-hoc reads: the extraction and copy reads on the held split (notebook C4)')
print('shipped', dest)
```
